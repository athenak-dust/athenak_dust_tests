## How to compile and run Athenak on Vista

Compile:

```bash
module purge
module load cmake
module load nvidia/24.7
module load openmpi/5.0.5

export NVCC_WRAPPER_DEFAULT_COMPILER=mpicxx
cmake -S athenak -B athenak/build-gpu-mpi \
  -DCMAKE_BUILD_TYPE=Release \
  -DPROBLEM="problem" \
  -DAthena_ENABLE_MPI=ON \
  -DKokkos_ENABLE_CUDA=ON \
  -DKokkos_ARCH_HOPPER90=ON \
  -DCMAKE_CXX_COMPILER="$PWD/athenak/kokkos/bin/nvcc_wrapper"
cmake --build athenak/build-gpu-mpi -j
```

Sbatch:
```bash
#!/bin/bash
#SBATCH --job-name=particle_scaling_02
#SBATCH --output=%j.out
#SBATCH --error=%j.err
#SBATCH --time=01:00:00
#SBATCH --nodes=1
#SBATCH --ntasks-per-node=1
#SBATCH --partition=gb
#SBATCH --account="Account"

cd "dir"

module purge
module load nvidia/24.7
module load openmpi/5.0.5
```

If errors with mpi occur with different module versions, try:

```bash
export UCX_CUDA_COPY_DMABUF=no
```

or

```bash
export UCX_TLS=^gdr_copy
```

This may affect performance.
